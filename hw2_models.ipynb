```python
from sklearn.metrics import classification_report
from sklearn.model_selection import LeaveOneGroupOut,cross_val_predict
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier 
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.neighbors import KNeighborsClassifier
import pandas as pd
import numpy as np
```

```python
#speed up
from sklearnex import patch_sklearn
patch_sklearn()  #cpu have to be intel
```

```python
data=pd.read_csv("process_dataSet.csv",index_col=0)
features=data.drop(["label","subject"],axis=1)
label=data["label"]
subject=data["subject"]
```

```python
#RandomForest
LOSO=LeaveOneGroupOut()
y_pred=cross_val_predict(RandomForestClassifier(),features,label,cv=LOSO,groups=subject)
report=classification_report(label,y_pred,digits=4)
print(report)
```

```python
#DecisionTree
LOSO=LeaveOneGroupOut()
y_pred=cross_val_predict(DecisionTreeClassifier(),features,label,cv=LOSO,groups=subject)
report=classification_report(label,y_pred,digits=4)
print(report)
```

```python
#AdaBoost
LOSO=LeaveOneGroupOut()
y_pred=cross_val_predict(AdaBoostClassifier(),features,label,cv=LOSO,groups=subject)
report=classification_report(label,y_pred,digits=4)
print(report)
```

```python
#LDA
LOSO=LeaveOneGroupOut()
y_pred=cross_val_predict(LinearDiscriminantAnalysis(),features,label,cv=LOSO,groups=subject)
report=classification_report(label,y_pred,digits=4)
print(report)
```

```python
#KNN
LOSO=LeaveOneGroupOut()
y_pred=cross_val_predict(KNeighborsClassifier(),features,label,cv=LOSO,groups=subject)
report=classification_report(label,y_pred,digits=4)
print(report)
```
